```python


# =========================
# STEP 2: Load Data
# =========================
# reload to ensure fresh state (safe in notebook)
df = pd.read_csv("f1_results_2025.csv")

# Ensure race_date is datetime
df["race_date"] = pd.to_datetime(df.get("race_date", pd.Series()), errors="coerce")

# =========================
# STEP 3: Feature Engineering
# =========================
def build_features(df):
    features = []
    # require columns
    req = ["driver_name", "race_date", "final_position", "status", "grid_position", "race_name"]
    for c in req:
        if c not in df.columns:
            raise KeyError(f"Required column missing: {c}")

    for driver in df["driver_name"].dropna().unique():
        driver_df = df[df["driver_name"] == driver].sort_values("race_date")
        # Need at least 5 prior races; first index with 5-history available is i=5 (0-based)
        for i in range(5, len(driver_df)):
            past_races = driver_df.iloc[i-5:i]
            if len(past_races) < 5:
                continue
            race = driver_df.iloc[i]

            driver_form = past_races["final_position"].astype(float).mean()
            driver_dnf_rate = (past_races["status"].str.contains("DNF", case=False, na=False)).mean()

            features.append({
                "driver_name": race["driver_name"],
                "race_name": race["race_name"],
                "race_date": race["race_date"],
                "grid_position": pd.to_numeric(race.get("grid_position", np.nan), errors="coerce"),
                "driver_form": driver_form,
                "driver_dnf_rate": float(driver_dnf_rate),
                "target_win": 1 if int(race.get("final_position", 999)) == 1 else 0,
                "target_podium": 1 if int(race.get("final_position", 999)) <= 3 else 0
            })
    return pd.DataFrame(features)

feat_df = build_features(df)

if feat_df.empty:
    print("No feature rows created (not enough history). Check races per driver.")
else:
    print(f"Built {len(feat_df)} training rows.")

# =========================
# STEP 4: Train Models
# =========================
def train_models(feat_df):
    X = feat_df[["grid_position", "driver_form", "driver_dnf_rate"]].copy()
    y_win = feat_df["target_win"].copy()
    y_pod = feat_df["target_podium"].copy()

    # Use index-based split so both targets/models share the exact same train/test rows
    idx_train, idx_test = train_test_split(feat_df.index, test_size=0.2, random_state=42, stratify=y_win if y_win.nunique()>1 else None)

    X_train = X.loc[idx_train]
    X_test = X.loc[idx_test]

    y_win_train = y_win.loc[idx_train]
    y_win_test = y_win.loc[idx_test]

    y_pod_train = y_pod.loc[idx_train]
    y_pod_test = y_pod.loc[idx_test]

    # Fill or impute simple NaNs (median for grid, mean for continuous)
    grid_median = X_train['grid_position'].median()
    X_train['grid_position'] = X_train['grid_position'].fillna(grid_median)
    X_test['grid_position'] = X_test['grid_position'].fillna(grid_median)

    form_mean = X_train['driver_form'].mean()
    X_train['driver_form'] = X_train['driver_form'].fillna(form_mean)
    X_test['driver_form'] = X_test['driver_form'].fillna(form_mean)

    X_train['driver_dnf_rate'] = X_train['driver_dnf_rate'].fillna(0.0)
    X_test['driver_dnf_rate'] = X_test['driver_dnf_rate'].fillna(0.0)

    model_win = XGBClassifier(

                                use_label_encoder=False,
    eval_metric='logloss',
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    gamma=0.1,
    min_child_weight=1,
    reg_alpha=0.0,
    reg_lambda=1.0,
    scale_pos_weight=1.0,   # compute as (n_negative / n_positive) for imbalance
    random_state=42,
    n_jobs=-1,
    verbosity=1)
    model_win.fit(X_train, y_win_train)

    model_pod = XGBClassifier(
                                use_label_encoder=False,
    eval_metric='logloss',
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    gamma=0.1,
    min_child_weight=1,
    reg_alpha=0.0,
    reg_lambda=1.0,
    scale_pos_weight=1.0,   # compute as (n_negative / n_positive) for imbalance
    random_state=42,
    n_jobs=-1,
    verbosity=1)
    model_pod.fit(X_train, y_pod_train)

    print("Winner model accuracy:", accuracy_score(y_win_test, model_win.predict(X_test)))
    print("Podium model accuracy:", accuracy_score(y_pod_test, model_pod.predict(X_test)))

    return model_win, model_pod

model_win, model_pod = (None, None)
if not feat_df.empty:
    model_win, model_pod = train_models(feat_df)

# =========================
# STEP 5: Predict Future Race
# =========================
def predict_future_race(df, race_name, race_date, model_win, model_pod, grid_dict=None, grid_weight: float = 0.3):
    """
    Predict per-driver win & podium probabilities for a future race.
    grid_weight controls how strongly qualifying position shifts win odds (pole highest).
    """
    race_date = pd.to_datetime(race_date)
    past_df = df[df["race_date"] < race_date]

    drivers = past_df["driver_name"].dropna().unique()
    features = []

    for driver in drivers:
        driver_history = past_df[past_df["driver_name"] == driver].sort_values("race_date").tail(5)
        if driver_history.empty:
            continue

        driver_form = driver_history["final_position"].astype(float).mean()
        driver_dnf_rate = (driver_history["status"].str.contains("DNF", case=False, na=False)).mean()

        if grid_dict and driver in grid_dict:
            grid_position = grid_dict[driver]
        else:
            grid_position = driver_history["grid_position"].dropna().mean()

        features.append({
            "driver_name": driver,
            "driver_form": driver_form,
            "driver_dnf_rate": float(driver_dnf_rate),
            "grid_position": float(grid_position) if not pd.isna(grid_position) else np.nan,
        })

    feat_df = pd.DataFrame(features)
    if feat_df.empty:
        print("No prediction rows - check past data and race_date")
        return feat_df

    # simple imputation
    feat_df['grid_position'] = feat_df['grid_position'].fillna(feat_df['grid_position'].median())
    feat_df['driver_form'] = feat_df['driver_form'].fillna(feat_df['driver_form'].median())
    feat_df['driver_dnf_rate'] = feat_df['driver_dnf_rate'].fillna(0.0)

    X = feat_df[["grid_position", "driver_form", "driver_dnf_rate"]]

    if model_win is not None:
        feat_df["win_prob"] = model_win.predict_proba(X)[:, 1]
    else:
        feat_df["win_prob"] = np.nan

    if model_pod is not None:
        feat_df["podium_prob"] = model_pod.predict_proba(X)[:, 1]
    else:
        feat_df["podium_prob"] = np.nan

    # Normalize and apply grid-based prior to win probabilities
    try:
        feat_df['win_prob_raw'] = feat_df['win_prob']
        eps = 1e-6
        p = feat_df['win_prob_raw'].clip(eps, 1 - eps)
        logits = np.log(p / (1 - p))
        # Grid bonus: pole (1) gets +grid_weight, P2 gets +(grid_weight * (1 - 1)), etc.
        # Using linear decay: bonus = -grid_weight * (grid_position - 1)
        grid_bonus = -grid_weight * (feat_df['grid_position'] - 1.0)
        adj_logits = logits + grid_bonus.values
        exps = np.exp(adj_logits - adj_logits.max())
        feat_df['win_prob'] = exps / exps.sum()
        feat_df['win_prob_norm'] = feat_df['win_prob']
        feat_df.drop('win_prob_raw', axis=1, inplace=True)
        print(f"Sum of win_prob after grid-adjusted normalization: {feat_df['win_prob'].sum():.6f}")
    except Exception as e:
        print("Warning: failed to normalize with grid prior:", e)

    return feat_df.sort_values("win_prob", ascending=False)

# =========================
# STEP 6: Example - Predict Hungarian GP 2025 (adjust as needed)
# =========================
if model_win is not None:
    preds = predict_future_race(df, "Italian GP", "2025-10-27", model_win, model_pod, grid_dict=None, grid_weight=0.5)
    print("\nTop Predictions for Italian GP 2025 (grid-weighted):")
    print(preds.head())
else: 
    print("Models not trained; skipping example prediction.")
```

```python
# =========================
# STEP 7: PyTorch Model (5 Hidden Layers, Adam) for Win & Podium
# =========================
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from typing import Tuple

# Reuse feat_df built earlier; if empty, skip
if 'feat_df' not in globals() or feat_df.empty:
    print("feat_df missing or empty; build features first before training PyTorch model.")
else:
    # Select feature columns and targets
    torch_features = feat_df[["grid_position", "driver_form", "driver_dnf_rate"]].copy()
    torch_targets_win = feat_df["target_win"].astype(float).values
    torch_targets_pod = feat_df["target_podium"].astype(float).values

    # Basic numeric scaling (standardization)
    feat_mean = torch_features.mean()
    feat_std = torch_features.std().replace(0, 1.0)
    torch_features_std = (torch_features - feat_mean) / feat_std

    class RaceDataset(Dataset):
        def __init__(self, X: np.ndarray, y_win: np.ndarray, y_pod: np.ndarray):
            self.X = torch.tensor(X, dtype=torch.float32)
            self.y_win = torch.tensor(y_win, dtype=torch.float32).unsqueeze(1)
            self.y_pod = torch.tensor(y_pod, dtype=torch.float32).unsqueeze(1)
        def __len__(self):
            return self.X.shape[0]
        def __getitem__(self, idx):
            return self.X[idx], self.y_win[idx], self.y_pod[idx]

    # Train/validation split (80/20)
    idx = np.arange(len(torch_features_std))
    rng = np.random.default_rng(42)
    rng.shuffle(idx)
    split = int(0.8 * len(idx))
    idx_train, idx_val = idx[:split], idx[split:]

    X_train_np = torch_features_std.iloc[idx_train].values
    X_val_np = torch_features_std.iloc[idx_val].values
    y_train_win = torch_targets_win[idx_train]
    y_val_win = torch_targets_win[idx_val]
    y_train_pod = torch_targets_pod[idx_train]
    y_val_pod = torch_targets_pod[idx_val]

    train_ds = RaceDataset(X_train_np, y_train_win, y_train_pod)
    val_ds = RaceDataset(X_val_np, y_val_win, y_val_pod)

    train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)
    val_loader = DataLoader(val_ds, batch_size=128, shuffle=False)

    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    class RaceNet(nn.Module):
        def __init__(self, in_dim: int):
            super().__init__()
            # 5 hidden layers (add activations if needed)
            self.net = nn.Sequential(
                nn.Linear(in_dim, 64),
                nn.ReLU(),
                nn.Linear(64, 128),
                nn.ReLU(),
                nn.Linear(128, 256),
                nn.ReLU(),
                nn.Linear(256, 512),
                nn.ReLU(),
                nn.Linear(512, 16),
                nn.ReLU(),
            )
            # Dual heads
            self.head_win = nn.Linear(16, 1)
            self.head_pod = nn.Linear(16, 1)
        def forward(self, x):
            h = self.net(x)
            return self.head_win(h), self.head_pod(h)

    model_torch = RaceNet(in_dim=torch_features_std.shape[1]).to(device)

    # Loss & optimizer
    bce = nn.BCEWithLogitsLoss()
    optimizer = torch.optim.Adam(model_torch.parameters(), lr=1e-3, weight_decay=1e-4)

    def evaluate(model: nn.Module, loader: DataLoader) -> Tuple[float, float, float, float]:
        model.eval()
        total = 0
        win_correct = 0
        pod_correct = 0
        win_loss_cum = 0.0
        pod_loss_cum = 0.0
        with torch.no_grad():
            for Xb, yw, yp in loader:
                Xb, yw, yp = Xb.to(device), yw.to(device), yp.to(device)
                logit_w, logit_p = model(Xb)
                loss_w = bce(logit_w, yw)
                loss_p = bce(logit_p, yp)
                pw = torch.sigmoid(logit_w)
                pp = torch.sigmoid(logit_p)
                win_correct += ((pw > 0.5) == (yw > 0.5)).sum().item()
                pod_correct += ((pp > 0.5) == (yp > 0.5)).sum().item()
                total += Xb.size(0)
                win_loss_cum += loss_w.item() * Xb.size(0)
                pod_loss_cum += loss_p.item() * Xb.size(0)
        return (win_loss_cum / total, pod_loss_cum / total,
                win_correct / total, pod_correct / total)

    epochs = 80
    best_val = float('inf')
    patience = 10
    no_improve = 0
    for epoch in range(1, epochs + 1):
        model_torch.train()
        for Xb, yw, yp in train_loader:
            Xb, yw, yp = Xb.to(device), yw.to(device), yp.to(device)
            optimizer.zero_grad()
            logit_w, logit_p = model_torch(Xb)
            loss_w = bce(logit_w, yw)
            loss_p = bce(logit_p, yp)
            loss = loss_w + loss_p
            loss.backward()
            optimizer.step()
        win_l_val, pod_l_val, win_acc_val, pod_acc_val = evaluate(model_torch, val_loader)
        val_score = win_l_val + pod_l_val
        if epoch % 5 == 0 or epoch == 1:
            print(f"Epoch {epoch:03d} | Val Loss W/P: {win_l_val:.4f}/{pod_l_val:.4f} | Acc W/P: {win_acc_val:.3f}/{pod_acc_val:.3f}")
        if val_score < best_val - 1e-4:
            best_val = val_score
            no_improve = 0
            best_state = {k: v.cpu().clone() for k, v in model_torch.state_dict().items()}
        else:
            no_improve += 1
            if no_improve >= patience:
                print("Early stopping.")
                break
    # Load best
    if 'best_state' in locals():
        model_torch.load_state_dict(best_state)

    # Store scaler params for inference
    torch_scaler = {"mean": feat_mean, "std": feat_std}

    def predict_future_race_torch(df, race_name, race_date, model, scaler, grid_dict=None, grid_weight: float = 0.3):
        """
        Torch inference mirroring sklearn flow with grid_weight boosting pole and decaying down the grid.
        """
        race_date = pd.to_datetime(race_date)
        past_df = df[df["race_date"] < race_date]
        drivers = past_df["driver_name"].dropna().unique()
        rows = []
        for driver in drivers:
            hist = past_df[past_df["driver_name"] == driver].sort_values("race_date").tail(5)
            if hist.empty:
                continue
            driver_form = hist["final_position"].astype(float).mean()
            driver_dnf_rate = (hist["status"].str.contains("DNF", case=False, na=False)).mean()
            grid_pos = grid_dict.get(driver) if grid_dict and driver in grid_dict else hist["grid_position"].dropna().mean()
            rows.append({
                "driver_name": driver,
                "grid_position": grid_pos,
                "driver_form": driver_form,
                "driver_dnf_rate": driver_dnf_rate,
            })
        if not rows:
            print("No drivers for prediction.")
            return pd.DataFrame()
        pred_df = pd.DataFrame(rows)
        # Impute
        pred_df['grid_position'] = pred_df['grid_position'].fillna(pred_df['grid_position'].median())
        pred_df['driver_form'] = pred_df['driver_form'].fillna(pred_df['driver_form'].median())
        pred_df['driver_dnf_rate'] = pred_df['driver_dnf_rate'].fillna(0.0)
        Xn = pred_df[["grid_position", "driver_form", "driver_dnf_rate"]]
        # Scale
        Xn_std = (Xn - scaler['mean']) / scaler['std'].replace(0, 1.0)
        Xt = torch.tensor(Xn_std.values, dtype=torch.float32).to(device)
        model.eval()
        with torch.no_grad():
            logit_w, logit_p = model(Xt)
            # Use raw logits for grid adjustment
            logit_w_np = logit_w.cpu().numpy().ravel()
            podium_prob = torch.sigmoid(logit_p).cpu().numpy().ravel()
        # Grid-adjusted logits and softmax normalization
        grid_bonus = -grid_weight * (pred_df['grid_position'].values - 1.0)
        adj_logits = logit_w_np + grid_bonus
        exps = np.exp(adj_logits - adj_logits.max())
        win_prob = exps / exps.sum()
        pred_df['win_prob'] = win_prob
        pred_df['podium_prob'] = podium_prob
        print(f"Sum of win_prob (torch, grid-adjusted) after normalization: {pred_df['win_prob'].sum():.6f}")
        return pred_df.sort_values('win_prob', ascending=False)

    # Example inference (Italian GP date same as earlier example)
    preds_torch = predict_future_race_torch(df, "Italian GP", "2025-10-20", model_torch, torch_scaler, grid_weight=0.5)
    print("Top (PyTorch) Italian GP Predictions:")
    print(preds_torch.head())
```

# Time-series Winner Prediction (RNN / LSTM / GRU / Transformer + Softmax)

This section reframes winner prediction as a **time-series / sequence** problem: for each race we build a fixed-length sequence of each driver’s last *K* races, then predict the winner as a **multi-class classification over the drivers in that race**.


Key idea:
- Input per sample (race): `X` has shape **[drivers, seq_len, features]**
- Model outputs **one logit per driver**
- We compute a **softmax over drivers** to get winner probabilities that sum to 1 for the race
- Training uses **cross-entropy loss** on the winner’s driver index


You can switch architectures by setting `ARCH` to one of: `"rnn"`, `"lstm"`, `"gru"`, `"transformer"`.

```python
import os
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from dataclasses import dataclass
from typing import Dict, List, Tuple

# -------------------------
# Config (tuned for better generalization)
# -------------------------
ARCH = "transformer"   # one of: "rnn", "lstm", "gru", "transformer"
SEQ_LEN = 10            # a bit more history helps
BATCH_SIZE = 32
EPOCHS = 80
LR = 1e-3
WEIGHT_DECAY = 1e-3
SEED = 42

torch.manual_seed(SEED)
np.random.seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
device

# -------------------------
# Load data (prefer multi-season for real time-series learning)
# -------------------------
DATA_CANDIDATES = [
    "f1_results_2017_2025.csv",
    "f1_results_2022_2025.csv",
    "f1_results_2025_with_sprints.csv",
    "f1_results_2025.csv",
 ]
data_path = next((p for p in DATA_CANDIDATES if os.path.exists(p)), None)
if data_path is None:
    raise FileNotFoundError(f"None of these files found: {DATA_CANDIDATES}")

df_all = pd.read_csv(data_path)
print("Loaded:", data_path, "rows=", len(df_all))

# -------------------------
# Standardize expected columns
# -------------------------
def pick_col(df: pd.DataFrame, *names: str) -> str:
    for n in names:
        if n in df.columns:
            return n
    raise KeyError(f"Missing columns; tried: {names}")

col_driver = pick_col(df_all, "driver_name", "driver", "Driver")
col_race   = pick_col(df_all, "race_name", "grand_prix", "Grand Prix")
col_date   = pick_col(df_all, "race_date", "date", "Date")
col_final  = pick_col(df_all, "final_position", "position", "Result")

col_grid   = next((c for c in ["grid_position", "grid", "Grid"] if c in df_all.columns), None)
col_points = next((c for c in ["points", "Points"] if c in df_all.columns), None)
col_status = next((c for c in ["status", "Status"] if c in df_all.columns), None)
col_round  = next((c for c in ["round", "Round"] if c in df_all.columns), None)
col_season = next((c for c in ["season", "year", "Year"] if c in df_all.columns), None)

work = df_all.copy()
work[col_driver] = work[col_driver].astype(str)
work[col_date] = pd.to_datetime(work[col_date], errors='coerce', dayfirst=True)
work[col_final] = pd.to_numeric(work[col_final], errors='coerce')
if col_grid:
    work[col_grid] = pd.to_numeric(work[col_grid], errors='coerce')
else:
    work["grid_position"] = np.nan
    col_grid = "grid_position"
if col_points:
    work[col_points] = pd.to_numeric(work[col_points], errors='coerce')
else:
    work["points"] = np.nan
    col_points = "points"
if col_status is None:
    work["status"] = ""
    col_status = "status"

# A stable race id for grouping; prefer (season, round) if present, else (date, race_name)
if col_season and col_round:
    work[col_season] = pd.to_numeric(work[col_season], errors='coerce')
    work[col_round] = pd.to_numeric(work[col_round], errors='coerce')
    work["race_id"] = work[col_season].astype('Int64').astype(str) + "_" + work[col_round].astype('Int64').astype(str)
else:
    work["race_id"] = work[col_date].dt.strftime('%Y-%m-%d').fillna('NA') + "_" + work[col_race].astype(str)

# Basic cleanup
work = work.dropna(subset=[col_driver, col_date, col_final, "race_id"]).copy()
work = work.sort_values([col_date, "race_id", col_driver]).reset_index(drop=True)

# Keep only races where we can identify a unique winner
winners = work.loc[work[col_final] == 1, ["race_id", col_driver]].drop_duplicates()
winner_counts = winners["race_id"].value_counts()
valid_races = winner_counts[winner_counts == 1].index
work = work[work["race_id"].isin(valid_races)].copy()

print("After filtering valid races:", work['race_id'].nunique(), "races |", len(work), "rows")

# -------------------------
# Driver vocabulary (learnable driver embedding boosts accuracy)
# -------------------------
driver_vocab = sorted(work[col_driver].dropna().astype(str).unique().tolist())
driver_to_id = {d: i for i, d in enumerate(driver_vocab)}
PAD_DRIVER_ID = len(driver_vocab)
print("Unique drivers:", len(driver_vocab))

# -------------------------
# Feature engineering for sequences
# -------------------------
def is_dnf(status: str) -> float:
    if not isinstance(status, str):
        return 0.0
    s = status.strip().lower()
    return 1.0 if ("dnf" in s or "ret" in s or "retired" in s) else 0.0

# Past-race features used inside the per-driver history sequence
work["feat_finish"] = work[col_final].astype(float)
work["feat_grid"] = work[col_grid].astype(float)
work["feat_points"] = work[col_points].astype(float)
work["feat_dnf"] = work[col_status].astype(str).map(is_dnf).astype(float)

FEAT_COLS = ["feat_grid", "feat_finish", "feat_points", "feat_dnf"]
FEAT_DIM = len(FEAT_COLS)

# -------------------------
# Build race samples
# - X_hist: [D, T, F] (history only)
# - tmask:  [D, T] (1=real, 0=pad) so Transformer ignores padding
# - cur_grid: [D] (current race grid position, known before race)
# - driver_ids: [D] ints for driver embedding
# - y: winner_idx (index into this race's driver list)
# -------------------------
@dataclass
class RaceSample:
    X_hist: np.ndarray
    tmask: np.ndarray
    driver_mask: np.ndarray
    cur_grid: np.ndarray
    driver_ids: np.ndarray
    winner_idx: int
    race_id: str
    race_date: pd.Timestamp
    drivers: List[str]

def build_samples(df: pd.DataFrame, seq_len: int, max_drivers: int = 20) -> List[RaceSample]:
    samples: List[RaceSample] = []
    race_meta = (df[["race_id", col_date]].drop_duplicates().sort_values(col_date))
    race_ids = race_meta["race_id"].tolist()

    # driver -> history list of (date, feat_vector)
    history: Dict[str, List[Tuple[pd.Timestamp, np.ndarray]]] = {}

    for rid in race_ids:
        race_rows = df[df["race_id"] == rid].copy()
        race_rows[col_driver] = race_rows[col_driver].astype(str)
        race_rows = race_rows.sort_values(col_driver)
        rdate = race_rows[col_date].iloc[0]
        drivers = race_rows[col_driver].tolist()

        wrow = race_rows[race_rows[col_final] == 1]
        if len(wrow) != 1:
            continue
        winner_name = str(wrow[col_driver].iloc[0])

        used_drivers = drivers[:max_drivers]
        D = max_drivers
        X = np.zeros((D, seq_len, FEAT_DIM), dtype=np.float32)
        tmask = np.zeros((D, seq_len), dtype=np.float32)
        driver_mask = np.zeros((D,), dtype=np.float32)
        cur_grid = np.full((D,), np.nan, dtype=np.float32)
        driver_ids = np.full((D,), PAD_DRIVER_ID, dtype=np.int64)

        # current grid values for this race
        rr = race_rows.set_index(col_driver)
        for i, drv in enumerate(used_drivers):
            driver_mask[i] = 1.0
            driver_ids[i] = driver_to_id.get(drv, PAD_DRIVER_ID)
            if drv in rr.index:
                cur_grid[i] = float(rr.loc[drv, col_grid]) if pd.notna(rr.loc[drv, col_grid]) else np.nan

            past = history.get(drv, [])
            past_feats = [fv for (_d, fv) in past][-seq_len:]
            k = len(past_feats)
            if k > 0:
                # left pad zeros; real history on the right
                X[i, -k:, :] = np.stack(past_feats, axis=0)
                tmask[i, -k:] = 1.0
            # else: all-zero + all-pad mask (transformer will ignore)

        if winner_name in used_drivers:
            winner_idx = used_drivers.index(winner_name)
            samples.append(RaceSample(
                X_hist=X,
                tmask=tmask,
                driver_mask=driver_mask,
                cur_grid=cur_grid,
                driver_ids=driver_ids,
                winner_idx=winner_idx,
                race_id=rid,
                race_date=rdate,
                drivers=used_drivers,
            ))

        # update history after constructing sample (avoid leakage)
        for _, row in race_rows.iterrows():
            drv = str(row[col_driver])
            fv = row[FEAT_COLS].to_numpy(dtype=np.float32)
            history.setdefault(drv, []).append((rdate, fv))

    return samples

samples = build_samples(work, seq_len=SEQ_LEN, max_drivers=20)
samples = sorted(samples, key=lambda s: s.race_date)
print("Built samples:", len(samples))
print("Example shapes:", samples[0].X_hist.shape, samples[0].tmask.shape, samples[0].cur_grid.shape)

# -------------------------
# Time-based split: last 20% races as test
# -------------------------
n = len(samples)
n_test = max(1, int(0.2 * n))
train_samples = samples[:-n_test]
test_samples = samples[-n_test:]
print("Train races:", len(train_samples), "Test races:", len(test_samples))

# -------------------------
# Scale history features on train only
# -------------------------
def fit_hist_scaler(train_samples: List[RaceSample]) -> Tuple[np.ndarray, np.ndarray]:
    all_feats = []
    for s in train_samples:
        # only real timesteps contribute
        m = (s.tmask > 0).reshape(-1)
        A = s.X_hist.reshape(-1, FEAT_DIM)
        if m.any():
            all_feats.append(A[m])
    A = np.concatenate(all_feats, axis=0)
    mean = A.mean(axis=0)
    std = A.std(axis=0)
    std = np.where(std < 1e-6, 1.0, std)
    return mean.astype(np.float32), std.astype(np.float32)

hist_mean, hist_std = fit_hist_scaler(train_samples)
print("Hist scaler mean:", hist_mean, "std:", hist_std)

def apply_hist_scaler(samples: List[RaceSample], mean: np.ndarray, std: np.ndarray) -> None:
    for s in samples:
        s.X_hist = (s.X_hist - mean[None, None, :]) / std[None, None, :]

apply_hist_scaler(train_samples, hist_mean, hist_std)
apply_hist_scaler(test_samples, hist_mean, hist_std)

# -------------------------
# Scale current grid separately (train only)
# -------------------------
def fit_grid_scaler(train_samples: List[RaceSample]) -> Tuple[float, float]:
    vals = []
    for s in train_samples:
        m = s.driver_mask > 0
        g = s.cur_grid[m]
        g = g[np.isfinite(g)]
        if len(g):
            vals.append(g)
    if not vals:
        return 0.0, 1.0
    v = np.concatenate(vals, axis=0)
    mu = float(v.mean())
    sd = float(v.std())
    if sd < 1e-6:
        sd = 1.0
    return mu, sd

grid_mean, grid_std = fit_grid_scaler(train_samples)
print("Grid mean/std:", grid_mean, grid_std)

def apply_grid_scaler(samples: List[RaceSample], mean: float, std: float) -> None:
    for s in samples:
        # impute per-race with mean then scale
        g = s.cur_grid.copy()
        g = np.where(np.isfinite(g), g, mean)
        s.cur_grid = ((g - mean) / std).astype(np.float32)

apply_grid_scaler(train_samples, grid_mean, grid_std)
apply_grid_scaler(test_samples, grid_mean, grid_std)
```

```python
from torch.utils.data import Dataset, DataLoader
from typing import Dict

class RaceWinnerDataset(Dataset):
    def __init__(self, samples: List[RaceSample]):
        self.samples = samples
    def __len__(self) -> int:
        return len(self.samples)
    def __getitem__(self, idx: int):
        s = self.samples[idx]
        return (
            torch.tensor(s.X_hist, dtype=torch.float32),      # [D, T, F]
            torch.tensor(s.tmask, dtype=torch.float32),       # [D, T]
            torch.tensor(s.driver_mask, dtype=torch.float32), # [D]
            torch.tensor(s.cur_grid, dtype=torch.float32),    # [D]
            torch.tensor(s.driver_ids, dtype=torch.long),     # [D]
            torch.tensor(s.winner_idx, dtype=torch.long),     # scalar
        )

def collate(batch):
    Xh, tm, dm, cg, did, y = zip(*batch)
    return (
        torch.stack(Xh, dim=0),   # [B, D, T, F]
        torch.stack(tm, dim=0),   # [B, D, T]
        torch.stack(dm, dim=0),   # [B, D]
        torch.stack(cg, dim=0),   # [B, D]
        torch.stack(did, dim=0),  # [B, D]
        torch.stack(y, dim=0),    # [B]
    )

train_loader = DataLoader(RaceWinnerDataset(train_samples), batch_size=BATCH_SIZE, shuffle=True, collate_fn=collate)
test_loader = DataLoader(RaceWinnerDataset(test_samples), batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate)

# -------------------------
# Models
# -------------------------
class PositionalEncoding(nn.Module):
    def __init__(self, d_model: int, max_len: int = 512):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2, dtype=torch.float32) * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        self.register_buffer('pe', pe.unsqueeze(0))  # [1, max_len, d_model]
    def forward(self, x: torch.Tensor) -> torch.Tensor:
        T = x.size(1)
        return x + self.pe[:, :T, :]

class DriverSequenceEncoder(nn.Module):
    def __init__(self, feat_dim: int, arch: str, hidden: int = 96, d_model: int = 96, nhead: int = 4, nlayers: int = 3, dropout: float = 0.15):
        super().__init__()
        self.arch = arch.lower()
        if self.arch in {"rnn", "lstm", "gru"}:
            rnn_cls = {"rnn": nn.RNN, "lstm": nn.LSTM, "gru": nn.GRU}[self.arch]
            self.rnn = rnn_cls(input_size=feat_dim, hidden_size=hidden, num_layers=2, batch_first=True, dropout=dropout)
            self.out_dim = hidden
        elif self.arch == "transformer":
            self.in_proj = nn.Linear(feat_dim, d_model)
            self.pos = PositionalEncoding(d_model=d_model, max_len=1024)
            enc_layer = nn.TransformerEncoderLayer(
                d_model=d_model, nhead=nhead, dim_feedforward=4*d_model, dropout=dropout, batch_first=True, norm_first=True
            )
            self.encoder = nn.TransformerEncoder(enc_layer, num_layers=nlayers)
            self.out_dim = d_model
        else:
            raise ValueError(f"Unknown arch: {arch}")

    def forward(self, x: torch.Tensor, tmask: torch.Tensor) -> torch.Tensor:
        """
        x: [N, T, F]
        tmask: [N, T] 1=real, 0=pad (left padded)
        returns: [N, H]
        """
        if self.arch in {"rnn", "lstm", "gru"}:
            out, _ = self.rnn(x)
            return out[:, -1, :]
        else:
            z = self.in_proj(x)
            z = self.pos(z)
            # Transformer expects True for positions that should be ignored
            key_padding = (tmask <= 0.0)  # [N, T]
            z = self.encoder(z, src_key_padding_mask=key_padding)
            return z[:, -1, :]

class RaceWinnerModel(nn.Module):
    def __init__(self, feat_dim: int, arch: str, n_drivers: int, driver_emb_dim: int = 24):
        super().__init__()
        self.encoder = DriverSequenceEncoder(feat_dim=feat_dim, arch=arch)
        self.driver_emb = nn.Embedding(n_drivers + 1, driver_emb_dim, padding_idx=n_drivers)
        in_dim = self.encoder.out_dim + driver_emb_dim + 1  # + current grid scalar
        self.mlp = nn.Sequential(
            nn.Linear(in_dim, 128),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(128, 1),
        )
        self.pad_driver_id = n_drivers

    def forward(self, X_hist: torch.Tensor, tmask: torch.Tensor, driver_mask: torch.Tensor, cur_grid: torch.Tensor, driver_ids: torch.Tensor) -> torch.Tensor:
        """
        X_hist: [B, D, T, F]
        tmask:  [B, D, T]
        driver_mask: [B, D]
        cur_grid: [B, D] (scaled)
        driver_ids: [B, D] (PAD id for padded drivers)
        returns logits: [B, D]
        """
        B, D, T, F = X_hist.shape
        Xf = X_hist.reshape(B * D, T, F)
        tf = tmask.reshape(B * D, T)
        emb_seq = self.encoder(Xf, tf)  # [B*D, H]
        did = driver_ids.reshape(B * D).clamp(min=0, max=self.pad_driver_id)
        emb_drv = self.driver_emb(did)  # [B*D, E]
        cg = cur_grid.reshape(B * D, 1)
        z = torch.cat([emb_seq, emb_drv, cg], dim=1)
        score = self.mlp(z).reshape(B, D)
        score = score.masked_fill(driver_mask <= 0.0, -1e9)
        return score

model = RaceWinnerModel(feat_dim=FEAT_DIM, arch=ARCH, n_drivers=len(driver_vocab)).to(device)
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
criterion = nn.CrossEntropyLoss(label_smoothing=0.05)

def topk_accuracy(logits: torch.Tensor, y: torch.Tensor, k: int = 1) -> float:
    topk = torch.topk(logits, k=k, dim=1).indices
    return (topk == y.unsqueeze(1)).any(dim=1).float().mean().item()

@torch.no_grad()
def evaluate_loader(model: nn.Module, loader: DataLoader) -> Dict[str, float]:
    model.eval()
    loss_sum = 0.0
    n = 0
    acc1_sum = 0.0
    acc3_sum = 0.0
    for Xh, tm, dm, cg, did, y in loader:
        Xh, tm, dm, cg, did, y = Xh.to(device), tm.to(device), dm.to(device), cg.to(device), did.to(device), y.to(device)
        logits = model(Xh, tm, dm, cg, did)
        loss = criterion(logits, y)
        b = Xh.size(0)
        loss_sum += loss.item() * b
        acc1_sum += topk_accuracy(logits, y, k=1) * b
        acc3_sum += topk_accuracy(logits, y, k=min(3, logits.size(1))) * b
        n += b
    return {"loss": loss_sum / max(1, n), "acc@1": acc1_sum / max(1, n), "acc@3": acc3_sum / max(1, n)}

# -------------------------
# Train loop (with proper early stopping)
# -------------------------
best_loss = float('inf')
best_state = None
patience = 12
no_improve = 0

for epoch in range(1, EPOCHS + 1):
    model.train()
    for Xh, tm, dm, cg, did, y in train_loader:
        Xh, tm, dm, cg, did, y = Xh.to(device), tm.to(device), dm.to(device), cg.to(device), did.to(device), y.to(device)
        optimizer.zero_grad()
        logits = model(Xh, tm, dm, cg, did)
        loss = criterion(logits, y)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

    train_metrics = evaluate_loader(model, train_loader)
    test_metrics = evaluate_loader(model, test_loader)

    if epoch == 1 or epoch % 5 == 0:
        print(f"Epoch {epoch:03d} | train loss={train_metrics['loss']:.4f} acc@1={train_metrics['acc@1']:.3f} | test loss={test_metrics['loss']:.4f} acc@1={test_metrics['acc@1']:.3f}")

    if test_metrics["loss"] < best_loss - 1e-4:
        best_loss = test_metrics["loss"]
        best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        no_improve = 0
    else:
        no_improve += 1
        if no_improve >= patience:
            print("Early stopping.")
            break

if best_state is not None:
    model.load_state_dict(best_state)

final_metrics = evaluate_loader(model, test_loader)
print("Final test metrics:", final_metrics)

# -------------------------
# Inference example: softmax winner probabilities for the last test race
# -------------------------
last = test_samples[-1]
Xh = torch.tensor(last.X_hist, dtype=torch.float32).unsqueeze(0).to(device)
tm = torch.tensor(last.tmask, dtype=torch.float32).unsqueeze(0).to(device)
dm = torch.tensor(last.driver_mask, dtype=torch.float32).unsqueeze(0).to(device)
cg = torch.tensor(last.cur_grid, dtype=torch.float32).unsqueeze(0).to(device)
did = torch.tensor(last.driver_ids, dtype=torch.long).unsqueeze(0).to(device)

model.eval()
with torch.no_grad():
    logits = model(Xh, tm, dm, cg, did)
    probs = torch.softmax(logits, dim=1).squeeze(0)
    probs_np = probs.detach().cpu().numpy()

pairs = [(drv, float(p)) for drv, p in zip(last.drivers, probs_np[:len(last.drivers)])]
pairs = sorted(pairs, key=lambda t: t[1], reverse=True)
print("Race:", last.race_id, "date:", str(last.race_date.date()))
print("Top-5 predicted winners (softmax probs):")
for drv, p in pairs[:5]:
    print(f"  {drv:>18s} : {p:.3f}")
print("True winner:", last.drivers[last.winner_idx])
```

## Notes / How to use

- Change `ARCH` to try different sequence models:
  - `ARCH = "rnn"` (vanilla RNN)
  - `ARCH = "lstm"`
  - `ARCH = "gru"`
  - `ARCH = "transformer"` (TransformerEncoder)

- The model outputs **logits per driver**; winner probabilities are computed with:
  - `probs = torch.softmax(logits, dim=1)` (softmax across drivers)

- If you only train on `f1_results_2025.csv`, the history is short. For better learning, keep the default multi-season load (`f1_results_2017_2025.csv`) so sequences have meaningful past context.
